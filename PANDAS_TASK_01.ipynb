#Manipulation of Tips Data


```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset
df = sns.load_dataset("tips")

# Display the dataset
df.head()
```

##Loading and Understanding Data

```python
# Check shape
print("Shape:", df.shape)

# View column names
print("Column Names:", df.columns.tolist())

# Check data types
print("\nData Types:\n", df.dtypes)

# Dataframe summary
print("\nDataFrame Info:")
df.info()
```

# Insights on Dataset Structure
Dimensions: The dataset consists of 244 rows and 7 columns.

Memory Efficiency: Features like sex, smoker, day, and time are stored as pandas category dtypes, which optimizes memory utilization.

Data Completeness: The dataset contains 244 non-null entries per column, confirming zero missing values out of the box.

Data Types: Consists of 2 floating-point numeric columns (total_bill, tip), 1 integer column (size), and 4 categorical columns.

##Data Exploration

```python
# First 5 rows
df.head()

# Descriptive statistics
df.describe()
```

### Insights on Numerical Patterns
Bill Scale: total_bill ranges from a minimum of $\$3.07$ to a maximum of $\$50.81$, with a mean of $\$19.78$.

Tip Characteristics: The mean tip is $\$3.00$, with values ranging from $\$1.00$ up to $\$10.00$.

Party Sizes: Party size (size) ranges from 1 to 6 people, with 50% of parties consisting of exactly 2 dining guests (median = 2).

Distribution Skewness: The mean total_bill ($\$19.78$) is higher than its median ($\$17.79$), indicating slight right-skewness driven by a few high-bill outliers.

##Indexing & Selection

```python
# 1. First 5 rows and columns 'total_bill', 'tip' using iloc
print(df.iloc[:5, [0, 1]])

# 2. Last 3 rows and columns 'day', 'time' using iloc
print( df.iloc[-3:, [4, 5]])

# 3. Index labels 0 to 10 and columns 'total_bill', 'sex' using loc
print( df.loc[0:10, ["total_bill", "sex"]])

# 4. Only 'tip' column for all rows using loc
print( df.loc[:, ["tip"]])

# 5. Row index 5 and columns 'total_bill', 'size' using loc
print(df.loc[5, ["total_bill", "size"]])

# 6. Position 10 to 20 and columns position 1 to 3 using iloc
print(df.iloc[10:20, 1:4])

# 7. First 10 rows and first 2 columns using iloc
print(df.iloc[:10, :2])

# 8. Rows where total_bill > 20 and columns 'total_bill', 'tip' using loc
print( df.loc[df["total_bill"] > 20, ["total_bill", "tip"]])

# 9. Rows where day == 'Sunday' and columns 'total_bill', 'tip', 'size' using loc
print(df.loc[df["day"] == "Sunday", ["total_bill", "tip", "size"]])
```

##Feature Creation


```python
# 1. Create tip_percentage feature
df["tip_percentage"] = (df["tip"] / df["total_bill"]) * 100

# 2. Binning total_bill using pd.cut
bill_labels = ["Low", "Medium", "High", "Very High"]
df["bill_tier"] = pd.cut(
    df["total_bill"], bins=[0, 12, 22, 35, 60], labels=bill_labels
)

df[["total_bill", "tip", "tip_percentage", "bill_tier"]].head()
```

##GroupBy Analysis

```python
# Group by gender
print(
    df.groupby("sex")[["total_bill", "tip", "tip_percentage"]].mean(
        numeric_only=True
    )
)

```

```python
# Group by day
print(
    df.groupby("day")[["total_bill", "tip", "tip_percentage"]].mean(
        numeric_only=True
    )
)

```

```python
# Group by time
print(
    df.groupby("time")[["total_bill", "tip", "tip_percentage"]].mean(
        numeric_only=True
    )
)
```

##Multi-Level Grouping

```python
# Multi-level group by day and time
multi_group = df.groupby(["day", "time"], observed=False)[
    ["total_bill", "tip"]
].agg(["mean", "count"])
multi_group
```

##Crosstab Analysis

```python
# Crosstab of Day vs Smoker status
ct_day_smoker = pd.crosstab(df["day"], df["smoker"], margins=True)
ct_day_smoker
```

## Pivot Tables

```python
# Pivot table: Mean tip percentage grouped by day and sex
pivot_tips = df.pivot_table(
    index="day", columns="sex", values="tip_percentage", aggfunc="mean"
)
pivot_tips
```

##Ranking Data

```python
# Rank based on bill and tip percentage
df["bill_rank"] = df["total_bill"].rank(ascending=False)
df["tip_pct_rank"] = df["tip_percentage"].rank(ascending=False)

df.sort_values(by="tip_pct_rank").head(5)
```

##Value Transformation

```python
# Map time to numeric codes
time_map = {"Lunch": 0, "Dinner": 1}
df["time_code"] = df["time"].map(time_map)

df[["time", "time_code"]].head()
```

##Category Diversity

```python
# Unique values and counts
for col in ["day", "time", "sex", "smoker", "size"]:
    print(
        f"{col} -> Unique Count: {df[col].nunique()}, Unique Values: {df[col].unique().tolist()}"
    )
```

##Handling Duplicates

```python
# Check duplicates
dup_count = df.duplicated().sum()
print("Duplicate Rows Count:", dup_count)

# Remove duplicates if present
df_no_dups = df.drop_duplicates()
print("Shape after removing duplicates:", df_no_dups.shape)
```

##Data Visualization Using Pandas

```python
# 1. Create tip_percentage feature if it doesn't exist (to avoid KeyError if cell was run out of order)
if 'tip_percentage' not in df.columns:
    df['tip_percentage'] = (df['tip'] / df['total_bill']) * 100

# 1. Bar Chart: Average Total Bill by Day
plt.figure(figsize=(6, 4))
df.groupby("day")["total_bill"].mean().plot(kind="bar", color="skyblue")
plt.title("Average Total Bill by Day")
plt.ylabel("Bill ($)")
plt.show()

# 2. Pie Chart: Proportion of Customers by Time
plt.figure(figsize=(5, 5))
df["time"].value_counts().plot(
    kind="pie", autopct="%1.1f%%", colors=["coral", "lightgreen"]
)
plt.title("Customer Distribution by Time")
plt.ylabel("")
plt.show()

# 3. Line Chart: Tip Percentage Across Sample Index
plt.figure(figsize=(8, 4))
df["tip_percentage"].plot(kind="line", color="purple")
plt.title("Tip Percentage Across Observations")
plt.xlabel("Index")
plt.ylabel("Tip %")
plt.show()

# 4. Scatter Plot: Total Bill vs Tip
plt.figure(figsize=(6, 4))
df.plot(kind="scatter", x="total_bill", y="tip", color="green")
plt.title("Total Bill vs Tip")
plt.show()

# 5. Stacked Bar Chart: Day vs Smoker
plt.figure(figsize=(6, 4))
pd.crosstab(df["day"], df["smoker"]).plot(
    kind="bar", stacked=True, colormap="viridis"
)
plt.title("Smoker vs Non-Smoker Distribution by Day")
plt.ylabel("Count")
plt.show()
```
